# Download 5 Mutual Funds

```python
import requests
import pandas as pd

codes = [
    119551,
    120503,
    118632,
    119092,
    120841
]

for code in codes:

    url = f"https://api.mfapi.in/mf/{code}"

    data = requests.get(url).json()

    df = pd.DataFrame(data["data"])

    df.to_csv(f"{code}.csv", index=False)

    print(f"{code} Downloaded")
```

# Insight:
# This code automates the retrieval of historical NAV data for five major mutual fund schemes using their AMFI codes. It connects to the MFAPI service, fetches the NAV history for each scheme, converts the JSON response into a structured DataFrame, and saves each dataset as a separate CSV file. This process enables efficient collection of mutual fund performance data for comparative analysis and trend evaluation
